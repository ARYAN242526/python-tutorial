### Pandas Series

A Pandas Series is like a column in a table. It is a 1-D array holding data of any type.

### Importing Pandas

```python
import numpy as np
import pandas as pd
```

### Series from lists

```python
# string
country = ['India', 'Australia', 'Nepal', 'USA', 'Sri Lanka']

s = pd.Series(country)
s
```

```python
# integers
runs = [13,24,56,78,100]

runs_ser = pd.Series(runs)
```

```python
# custom index
marks = [67,57,89,100]
subjects = ['maths','english','science','hindi']

pd.Series(marks,index=subjects)
```

```python
# setting a name
marks = pd.Series(marks,index=subjects,name='Aryan ke marks')
marks
```

### Series from dict

```python
marks = {
    'maths': 67,
    'english': 57,
    'science': 89,
    'hindi': 100
}

marks_series = pd.Series(marks,name='aryan ke marks')
marks_series
```
