### Creating Numpy Arrays

```python
# np array
import numpy as np

a = np.array([1,2,3])
print(a)
```

```python
# 2D and 3D
b= np.array([[1,2,3], [4,5,6]])
print(b)
```

```python
c = np.array([[[1,2], [3,4]],[[5,6], [7,8]]])
print(c)
```

```python
# dtype
np.array([1,2,3], dtype=float)
```

```python
# np.arange
np.arange(1,11)
```

```python
# with reshape
np.arange(16).reshape(2,2,2,2)
```

```python
# np.ones and np.zeros
np.ones((3,4))
```

```python
np.zeros((3,4))
```

```python
# np.random
np.random.random((3,4))
```

```python
# np.linspace
np.linspace(-10,10,10,dtype=int)
```

```python
# np.identity
np.identity(3)
```

### Array Attributes

```python
a1 = np.arange(10, dtype=np.int32)
a2 = np.arange(12, dtype=float).reshape(3,4)
a3 = np.arange(8).reshape(2,2,2)

a3
```

```python
# ndim
a3.ndim
```

```python
# shape
print(a3.shape)
a3
```

```python
# size
print(a2.size)
a2
```

```python
# itemsize
a3.itemsize
```

```python
# dtype

print(a1.dtype)
print(a2.dtype)
print(a3.dtype)
```

### Changing Datatype

```python
# astype
a3.astype(np.int32)
```

### Array Operations

```python
a1 = np.arange(12).reshape(3,4)
a2 = np.arange(12,24).reshape(3,4)

a2
```

```python
# scalar operations

# arithmetic
a1 ** 2
```

```python
# relational
a2 == 15
```

```python
# vector operations
# arithmetic
a1 ** a2
```

### Array Functions

```python
a1 = np.random.random((3,3))
a1 = np.round(a1 * 100)
a1
```

```python
# max/min/sum/prod
# 0 -> col and 1 -> row
np.prod(a1, axis=0)
```
